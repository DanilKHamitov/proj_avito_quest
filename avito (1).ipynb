```python
#загрузка данных
benchmark_queries = pd.read_parquet('/kaggle/input/datasets/khamitovdanil/data-for-avito/benchmark_queries.parquet')

benchmark_items = pd.read_parquet('/kaggle/input/datasets/khamitovdanil/data-for-avito/benchmark_items.parquet')

train = pd.read_parquet('/kaggle/input/datasets/khamitovdanil/data-for-avito/train.parquet')

print(f"Размер тестовых запросов: {benchmark_queries.shape}")
print(f"Размер базы объявлений: {benchmark_items.shape}")
print(f"Размер трейна: {train.shape}")
```

```python
#Здесь базово выведено для каждой таблицы пропуски, уникальные значения и типы данных чтобы понять, какую работу с данными надо проделать,что изменить и что не использовать.
import pandas as pd

def get_data_report(df, dataset_name):
    print(f"--- Сводка по датасету: {dataset_name} ---")
    print(f"Размер: {df.shape[0]} строк, {df.shape[1]} колонок\n")
    
    report = pd.DataFrame({
        'Тип данных': df.dtypes,
        'Пропусков (шт)': df.isna().sum(),
        'Пропусков (%)': (df.isna().sum() / len(df) * 100).round(2),
        'Уникальных значений': df.nunique()
    })
    
    display(report)
    

get_data_report(benchmark_queries, "Queries (Запросы)")
get_data_report(benchmark_items, "Items (Объявления)")
get_data_report(train, "Train (Обучающая выборка)")
```

```python
#работа с данными.
#склеил строки с текстовым описанием,чтобы можно было подавать текст в трансформер в удобном виде
# также были заменены 35 пропусков в item_description на пустые, так как при склейке с NaN получается Nan
benchmark_items['item_description_raw'] = benchmark_items['item_description_raw'].fillna('')
train['item_description_raw'] = train['item_description_raw'].fillna('')


benchmark_items['item_full_text'] = (
    benchmark_items['item_title_raw'] + " " + 
    benchmark_items['item_description_raw'] + " " + 
    benchmark_items['item_infm_params_text']
).str.lower()

 
benchmark_queries['query_full_text'] = (
    benchmark_queries['search_query'] + " " + 
    benchmark_queries['search_infm_params_text']
).str.lower()



for df in [benchmark_items, train]:
    df['item_rating_reviews_count'] = df['item_rating_reviews_count'].fillna(0)
    df['item_rating'] = df['item_rating'].fillna(-1)
    
  
    df['item_latitude'] = pd.to_numeric(df['item_latitude'], errors='coerce')
    df['item_longitude'] = pd.to_numeric(df['item_longitude'], errors='coerce')
    
  
    df['item_latitude'] = df.groupby('item_location_id')['item_latitude'].transform(lambda x: x.fillna(x.mean()))
    df['item_longitude'] = df.groupby('item_location_id')['item_longitude'].transform(lambda x: x.fillna(x.mean()))
    
   
    df['item_latitude'] = df['item_latitude'].fillna(0)
    df['item_longitude'] = df['item_longitude'].fillna(0)
```

```python
# здесь проверил совпадают  ли местополоожение и категория в train(чтобы понимать зависимость)
print((train['search_location_id'] == train['item_location_id']).mean())

print((train['search_category'] == train['item_category_id']).mean())
```

```python
import pandas as pd
from sentence_transformers import SentenceTransformer, InputExample, losses
from torch.utils.data import DataLoader
import torch

device = 'cuda' if torch.cuda.is_available() else 'cpu'

#Загружаем базу.Тексты макс 64 токена,иначе вылетает.
model = SentenceTransformer('cointegrated/rubert-tiny2', device=device)
model.max_seq_length = 64

# Берем весь трейн с перемешиванием
train_subset = train.sample(n=497673, random_state=42).reset_index(drop=True)

# Склеиваем заголовок и описание для более плотного контекста
queries = train_subset['search_query'].fillna('')
item_titles = train_subset['item_title_raw'].fillna('')
item_descriptions = train_subset['item_description_raw'].fillna('')
item_full_texts = item_titles + " " + item_descriptions

# Формируем пары (запрос -> кликнутое объявление)
train_examples = []
for q, text in zip(queries, item_full_texts):
    if len(q) > 0 and len(text) > 0: 
        train_examples.append(InputExample(texts=[str(q), str(text)]))

print(f"Собрано {len(train_examples)} пар для обучения. Запускаем...")

# Батч 64 — адекватный лимит под доступную VRAM
train_dataloader = DataLoader(train_examples, shuffle=True, batch_size=64)

# Используем in-batch negatives
train_loss = losses.MultipleNegativesRankingLoss(model=model)

#обучение на одной эпохе
model.fit(
    train_objectives=[(train_dataloader, train_loss)],
    epochs=1,         
    warmup_steps=100, 
    show_progress_bar=True,
    output_path='./rubert-tiny2-avito-tuned'
)
print("Обучение завершено! Модель сохранена.")
```

```python
from sentence_transformers import SentenceTransformer
import torch
import pandas as pd
from tqdm.notebook import tqdm

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Поднимаем нашу дообученную модель с диска
tuned_model = SentenceTransformer('./rubert-tiny2-avito-tuned', device=device)
tuned_model.max_seq_length = 64

print("Генерация новых векторов...")
#применение к items модели готовой
items_embeddings_tuned = tuned_model.encode(
    benchmark_items['item_full_text'].tolist(),
    batch_size=256, show_progress_bar=True, convert_to_tensor=True, normalize_embeddings=True
)
#применеие к queries модели готовой 
queries_embeddings_tuned = tuned_model.encode(
    benchmark_queries['query_full_text'].tolist(),
    batch_size=256, show_progress_bar=True, convert_to_tensor=True, normalize_embeddings=True
)

#перекидываем фичи локаций на GPU, чтобы быстро работало
item_locs = torch.tensor(benchmark_items['item_location_id'].values, device=device)
item_ids = benchmark_items['item_id'].values
predictions = []

print("Поиск кандидатов...")
for idx, query_row in tqdm(benchmark_queries.iterrows(), total=len(benchmark_queries)):
    q_vec = queries_embeddings_tuned[idx]
    q_loc = query_row['search_location_id']

    #косинусное расстояние считаем
    sim_scores = torch.mv(items_embeddings_tuned, q_vec)
    
    # объявления из одного города бустим
    sim_scores += (item_locs == q_loc).float() * 0.2

    #вытаскиваем топ 50 подходящих
    top_indices = torch.topk(sim_scores, k=50).indices.cpu().numpy()
    predictions.append(' '.join(item_ids[top_indices]))

answer = pd.DataFrame({
    'query_id': benchmark_queries['query_id'],
    'answer': predictions
})
answer.to_csv('answer_tuned.csv', index=False)
print("Файл answer_tuned.csv успешно создан!")
```
